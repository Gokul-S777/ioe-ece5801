```python
# Sensor Data Acquisition and Preprocessing using Python

import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------
# Step 1: Create Sensor Dataset
# ----------------------------
sensor_data = {
    "Temperature":[28,29,30,31,29],
    "Humidity":[65,66,68,69,67],
    "Pressure":[1012,1011,1010,1013,1011],
    "HeartRate":[78,80,81,79,80],
    "SpO2":[98,99,97,98,99]
}

df = pd.DataFrame(sensor_data)

# Save dataset
df.to_csv("sensor_data.csv", index=False)

# ----------------------------
# Step 2: Read Dataset
# ----------------------------
df = pd.read_csv("sensor_data.csv")

print("Original Dataset")
print(df)

# ----------------------------
# Step 3: Data Preprocessing
# ----------------------------

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Display cleaned dataset
print("\nCleaned Dataset")
print(df)

# Statistical Summary
print("\nStatistical Summary")
print(df.describe())

# ----------------------------
# Step 4: Data Visualization
# ----------------------------

# Temperature
plt.figure(figsize=(6,4))
plt.plot(df["Temperature"], marker='o')
plt.title("Temperature")
plt.xlabel("Sample")
plt.ylabel("Temperature (°C)")
plt.grid()
plt.show()

# Humidity
plt.figure(figsize=(6,4))
plt.plot(df["Humidity"], marker='o')
plt.title("Humidity")
plt.xlabel("Sample")
plt.ylabel("Humidity (%)")
plt.grid()
plt.show()

# Heart Rate
plt.figure(figsize=(6,4))
plt.bar(df.index, df["HeartRate"])
plt.title("Heart Rate")
plt.xlabel("Sample")
plt.ylabel("Heart Rate (BPM)")
plt.show()

# SpO2
plt.figure(figsize=(6,4))
plt.bar(df.index, df["SpO2"])
plt.title("SpO2")
plt.xlabel("Sample")
plt.ylabel("SpO2 (%)")
plt.show()

# Pressure
plt.figure(figsize=(6,4))
plt.plot(df["Pressure"], marker='o')
plt.title("Pressure")
plt.xlabel("Sample")
plt.ylabel("Pressure (hPa)")
plt.grid()
plt.show()

print("\nExperiment Completed Successfully!")
```
