```python
# ===========================================
# Case Study : Smart Healthcare Monitoring System
# Google Colab Simulation
# ===========================================

import random
import time
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import clear_output

# -------------------------
# Patient Names
# -------------------------

patients = [
    "Patient 1", "Patient 2", "Patient 3", "Patient 4", "Patient 5",
    "Patient 6", "Patient 7", "Patient 8", "Patient 9", "Patient 10"
]

respiratory_data = []
systolic_data = []
diastolic_data = []
glucose_data = []
pulse_data = []
status_data = []

# -------------------------
# Simulation
# -------------------------

for patient in patients:

    respiratory = random.randint(10, 25)
    systolic = random.randint(80, 160)
    diastolic = random.randint(50, 100)
    glucose = random.randint(60, 180)
    pulse = random.randint(50, 125)

    respiratory_data.append(respiratory)
    systolic_data.append(systolic)
    diastolic_data.append(diastolic)
    glucose_data.append(glucose)
    pulse_data.append(pulse)

    # -------------------------
    # Emergency Detection
    # -------------------------

    if (
        respiratory < 12 or respiratory > 20 or
        systolic < 90 or systolic > 140 or
        diastolic < 60 or diastolic > 90 or
        glucose < 70 or glucose > 140 or
        pulse < 60 or pulse > 100
    ):
        status = "EMERGENCY"
        color = "red"
    else:
        status = "NORMAL"
        color = "green"

    status_data.append(status)

    # -------------------------
    # Live Display
    # -------------------------

    clear_output(wait=True)

    print("=" * 65)
    print("       SMART HEALTHCARE MONITORING SYSTEM")
    print("=" * 65)

    print("\nPatient :", patient)
    print("Respiratory Rate :", respiratory, "breaths/min")
    print("Systolic BP :", systolic, "mmHg")
    print("Diastolic BP :", diastolic, "mmHg")
    print("Blood Glucose :", glucose, "mg/dL")
    print("Pulse Rate :", pulse, "BPM")
    print("Status :", status)

    # -------------------------
    # Graphs
    # -------------------------

    plt.figure(figsize=(15, 8))

    # Respiratory Rate
    plt.subplot(2, 3, 1)
    plt.bar(
        ["Respiratory"],
        [respiratory]
    )
    plt.ylim(5, 30)
    plt.title("Respiratory Rate")
    plt.ylabel("Breaths/min")

    # Systolic BP
    plt.subplot(2, 3, 2)
    plt.bar(
        ["Systolic"],
        [systolic]
    )
    plt.ylim(70, 170)
    plt.title("Systolic Blood Pressure")
    plt.ylabel("mmHg")

    # Diastolic BP
    plt.subplot(2, 3, 3)
    plt.bar(
        ["Diastolic"],
        [diastolic]
    )
    plt.ylim(40, 110)
    plt.title("Diastolic Blood Pressure")
    plt.ylabel("mmHg")

    # Blood Glucose
    plt.subplot(2, 3, 4)
    plt.bar(
        ["Glucose"],
        [glucose]
    )
    plt.ylim(40, 200)
    plt.title("Blood Glucose")
    plt.ylabel("mg/dL")

    # Pulse Rate
    plt.subplot(2, 3, 5)
    plt.bar(
        ["Pulse"],
        [pulse]
    )
    plt.ylim(40, 140)
    plt.title("Pulse Rate")
    plt.ylabel("BPM")

    # Status
    plt.subplot(2, 3, 6)
    plt.bar(
        ["Status"],
        [1],
        color=color
    )
    plt.ylim(0, 1.2)
    plt.title("Patient Status")
    plt.text(
        0,
        0.5,
        status,
        ha="center",
        fontsize=12
    )
    plt.xticks([])

    plt.tight_layout()
    plt.show()

    time.sleep(2)

# -------------------------
# Final Report
# -------------------------

report = pd.DataFrame({

    "Patient": patients,
    "Respiratory Rate": respiratory_data,
    "Systolic BP": systolic_data,
    "Diastolic BP": diastolic_data,
    "Blood Glucose": glucose_data,
    "Pulse Rate": pulse_data,
    "Status": status_data

})

print("\n" + "=" * 65)
print("                    FINAL REPORT")
print("=" * 65)

print(report)

# -------------------------
# Summary
# -------------------------

normal = len(
    report[report["Status"] == "NORMAL"]
)

emergency = len(
    report[report["Status"] == "EMERGENCY"]
)

print("\nNORMAL Patients    :", normal)
print("EMERGENCY Patients :", emergency)

# -------------------------
# Save CSV
# -------------------------

report.to_csv(
    "Smart_Healthcare_Report.csv",
    index=False
)

print("\nCSV File Saved Successfully")
print("File Name: Smart_Healthcare_Report.csv")
print("\nHealthcare Monitoring Completed Successfully!")
```
