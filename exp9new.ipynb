```python
# ============================================
# AI-Based Traffic Prediction
# Different Parameter: Number of Vehicles
# Google Colab
# ============================================

import matplotlib.pyplot as plt

print("=" * 50)
print(" AI-BASED TRAFFIC PREDICTION")
print("=" * 50)

# Time intervals
time = ["8 AM", "9 AM", "10 AM", "11 AM", "12 PM", "1 PM", "2 PM"]

# Actual number of vehicles
actual = [45, 52, 60, 68, 75, 82, 90]

# Predicted number of vehicles
predicted = [47, 50, 62, 65, 73, 85, 88]

# --------------------------------------------
# Calculate Prediction Error
# --------------------------------------------

errors = []

for i in range(len(actual)):
    errors.append(abs(actual[i] - predicted[i]))

average_error = sum(errors) / len(errors)

# --------------------------------------------
# Display Values
# --------------------------------------------

print("\nTraffic Prediction Data")
print("-" * 50)

for i in range(len(time)):
    print(
        time[i],
        ": Actual =", actual[i],
        "vehicles, Predicted =", predicted[i],
        "vehicles"
    )

# --------------------------------------------
# Performance
# --------------------------------------------

print("\nPrediction Performance")
print("-" * 50)

print(
    "Average Prediction Error =",
    round(average_error, 2),
    "vehicles"
)

# --------------------------------------------
# Graph 1: Actual vs Predicted
# --------------------------------------------

plt.figure(figsize=(8, 4))

plt.plot(
    time,
    actual,
    marker="o",
    label="Actual Vehicles"
)

plt.plot(
    time,
    predicted,
    marker="s",
    label="Predicted Vehicles"
)

plt.title("Actual vs Predicted Number of Vehicles")
plt.xlabel("Time")
plt.ylabel("Number of Vehicles")
plt.legend()
plt.grid(True)

plt.show()

# --------------------------------------------
# Graph 2: Prediction Error
# --------------------------------------------

plt.figure(figsize=(8, 4))

plt.bar(
    time,
    errors
)

plt.title("Traffic Prediction Error")
plt.xlabel("Time")
plt.ylabel("Error (Vehicles)")
plt.grid(axis="y")

plt.show()

# --------------------------------------------
# Final Result
# --------------------------------------------

print("\n" + "=" * 50)
print("Traffic Prediction Analysis Completed")
print("Average Error =", round(average_error, 2), "vehicles")
print("=" * 50)
```
