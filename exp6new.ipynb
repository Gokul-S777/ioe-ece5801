```python
# ======================================================
# THREAT MODELING & SECURITY HARDENING
# IoE SMART BANKING NETWORK SECURITY SYSTEM
# Google Colab
# ======================================================

import random
import hashlib
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 65)
print("     IoE SMART BANKING NETWORK SECURITY SYSTEM")
print("=" * 65)

# ------------------------------------------------------
# Secure User Authentication
# ------------------------------------------------------

username = "bank_user"
password = "SecureBank@456"

password_hash = hashlib.sha256(
    password.encode()
).hexdigest()

print("\nAuthentication Successful")
print("User ID       :", username)
print("Security Hash :", password_hash[:20], "...")

# ------------------------------------------------------
# IoE Banking Network Devices
# ------------------------------------------------------

device_ids = [
    "NODE001",
    "NODE002",
    "NODE003",
    "NODE004",
    "NODE005",
    "NODE006",
    "NODE007",
    "NODE008"
]

records = []

for device in device_ids:

    login_attempts = random.randint(1, 15)

    transaction_frequency = random.randint(5, 100)

    network_traffic = random.randint(100, 1000)

    risk_score = random.randint(10, 100)

    threat = random.choice([
        "NORMAL",
        "Unauthorized Login",
        "Suspicious Traffic",
        "Credential Attack",
        "Data Injection"
    ])

    # --------------------------------------------------
    # Security Decision
    # --------------------------------------------------

    if threat == "NORMAL" and risk_score < 60:
        status = "ALLOWED"
    else:
        status = "BLOCKED"

    records.append([
        device,
        login_attempts,
        transaction_frequency,
        network_traffic,
        risk_score,
        threat,
        status
    ])

# ------------------------------------------------------
# Display Network Security Report
# ------------------------------------------------------

df = pd.DataFrame(
    records,
    columns=[
        "Device ID",
        "Login Attempts",
        "Transaction Frequency",
        "Network Traffic (KB)",
        "Risk Score",
        "Threat",
        "Status"
    ]
)

print("\nIoE Banking Network Security Report\n")
print(df.to_string(index=False))

# ------------------------------------------------------
# Security Threat Summary
# ------------------------------------------------------

normal = len(
    df[df["Threat"] == "NORMAL"]
)

unauthorized = len(
    df[df["Threat"] == "Unauthorized Login"]
)

traffic = len(
    df[df["Threat"] == "Suspicious Traffic"]
)

credential = len(
    df[df["Threat"] == "Credential Attack"]
)

injection = len(
    df[df["Threat"] == "Data Injection"]
)

print("\n========== SECURITY ANALYSIS ==========")

print("Normal Devices       :", normal)
print("Unauthorized Login   :", unauthorized)
print("Suspicious Traffic   :", traffic)
print("Credential Attack    :", credential)
print("Data Injection       :", injection)

# ------------------------------------------------------
# Security Status Summary
# ------------------------------------------------------

allowed = len(
    df[df["Status"] == "ALLOWED"]
)

blocked = len(
    df[df["Status"] == "BLOCKED"]
)

print("\n========== ACCESS REPORT ==========")

print("Allowed Devices :", allowed)
print("Blocked Devices :", blocked)

# ------------------------------------------------------
# GRAPH 1
# Login Attempts
# ------------------------------------------------------

plt.figure(figsize=(9, 4))

plt.bar(
    df["Device ID"],
    df["Login Attempts"]
)

plt.title("Login Attempts by IoE Device")
plt.xlabel("IoE Device")
plt.ylabel("Number of Login Attempts")
plt.grid(axis="y")

plt.show()

# ------------------------------------------------------
# GRAPH 2
# Risk Score
# ------------------------------------------------------

plt.figure(figsize=(9, 4))

plt.plot(
    df["Device ID"],
    df["Risk Score"],
    marker="o"
)

plt.title("Security Risk Score")
plt.xlabel("IoE Device")
plt.ylabel("Risk Score")
plt.grid(True)

plt.show()

# ------------------------------------------------------
# GRAPH 3
# Security Status
# ------------------------------------------------------

plt.figure(figsize=(6, 4))

plt.bar(
    ["ALLOWED", "BLOCKED"],
    [allowed, blocked]
)

plt.title("Network Access Status")
plt.xlabel("Security Status")
plt.ylabel("Number of Devices")

plt.show()

# ------------------------------------------------------
# GRAPH 4
# Threat Distribution
# ------------------------------------------------------

plt.figure(figsize=(7, 7))

plt.pie(
    [
        normal,
        unauthorized,
        traffic,
        credential,
        injection
    ],
    labels=[
        "NORMAL",
        "Unauthorized Login",
        "Suspicious Traffic",
        "Credential Attack",
        "Data Injection"
    ],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("IoE Security Threat Distribution")

plt.show()

# ------------------------------------------------------
# GRAPH 5
# Network Traffic
# ------------------------------------------------------

plt.figure(figsize=(9, 4))

plt.bar(
    df["Device ID"],
    df["Network Traffic (KB)"]
)

plt.title("Network Traffic by IoE Device")
plt.xlabel("IoE Device")
plt.ylabel("Network Traffic (KB)")
plt.grid(axis="y")

plt.show()

# ------------------------------------------------------
# Final Security Summary
# ------------------------------------------------------

print("\n" + "=" * 65)
print("       IoE SECURITY MONITORING COMPLETED")
print("=" * 65)

print("Total Devices Monitored :", len(df))
print("Allowed Devices         :", allowed)
print("Blocked Devices         :", blocked)
print("Total Security Threats  :", len(df[df["Status"] == "BLOCKED"]))

print("\nSecurity Hardening Measures:")
print("1. Password Hashing Applied")
print("2. Login Attempts Monitored")
print("3. Network Traffic Analyzed")
print("4. Risk Score Calculated")
print("5. Suspicious Devices Blocked")
print("6. Security Threats Classified")

print("\nIoE Banking Network Protected Successfully")
print("=" * 65)
```
