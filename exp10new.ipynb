```python
# ============================================
# IoT-Based Smart Agriculture
# Automated Irrigation using Soil Moisture
# Google Colab
# ============================================

import random
import matplotlib.pyplot as plt

print("=" * 55)
print("   IoT-BASED SMART AGRICULTURE SYSTEM")
print("=" * 55)

# Soil moisture readings (%)
soil_moisture = []

# Number of readings
readings = 10

# --------------------------------------------
# Soil Moisture Monitoring
# --------------------------------------------

for i in range(1, readings + 1):

    moisture = random.randint(20, 80)
    soil_moisture.append(moisture)

    # Irrigation control
    if moisture < 40:
        irrigation = "ON"
    else:
        irrigation = "OFF"

    print(
        "Reading", i,
        "| Soil Moisture =", moisture, "%",
        "| Irrigation =", irrigation
    )

# --------------------------------------------
# Graph
# --------------------------------------------

reading_numbers = list(range(1, readings + 1))

plt.figure(figsize=(9, 5))

plt.plot(
    reading_numbers,
    soil_moisture,
    marker="o"
)

# Moisture threshold
plt.axhline(
    40,
    linestyle="--",
    label="Irrigation Threshold"
)

plt.title("Soil Moisture Monitoring")
plt.xlabel("Reading Number")
plt.ylabel("Soil Moisture (%)")
plt.legend()
plt.grid(True)

plt.show()

# --------------------------------------------
# Final Summary
# --------------------------------------------

irrigation_on = sum(
    1 for value in soil_moisture if value < 40
)

irrigation_off = readings - irrigation_on

print("\n" + "=" * 55)
print("           IRRIGATION SUMMARY")
print("=" * 55)

print("Total Readings       :", readings)
print("Irrigation ON        :", irrigation_on)
print("Irrigation OFF       :", irrigation_off)

print("\nSmart Irrigation Monitoring Completed Successfully!")
print("=" * 55)
```
